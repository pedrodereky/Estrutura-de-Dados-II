## 15. Exercício 1 — Referências

Execute:


````python
a = [10, 20, 30]
b = a.copy()

b.append(40)

print("a:", a)
print("b:", b)
print("id(a):", id(a))
print("id(b):", id(b))
````

Responda:

1. Qual será a saída?

\
a: [10, 20, 30, 40]

b: [10, 20, 30, 40]

id(a): 138852283393600

id(b): 138852283393600

\
2. `a` e `b` representam o mesmo objeto?

R: Sim, `a` e `b` representam o mesmo objeto.

\
3. Por quê?

R: a e b representam o mesmo objeto porque ambas as variáveis **referenciam** (apontam para) o mesmo objeto.

\
4. O que aconteceria com `b = a.copy()`?

R: Nesse caso as listas não irão mostrar elementos identicos e nem o mesmo ID, pois listas são mutáveis, então uma copia vai criar um novo registro na memória.

## 16. Exercício 2 — Construindo uma cadeia
Crie:

```python
n1 = Node("A")
n2 = Node("B")
n3 = Node("C")
```

Conecte:

```text
A → B → C → None
```

Depois percorra a estrutura e produza:

```text
A
B
C
```

\
Código:

````python
class Node:

    def __init__(self, valor):
        self.valor = valor
        self.proximo = None

n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
    print(atual.valor)
    atual = atual.proximo
````

## 18. Problema real — Sistema de atendimento de uma clínica

Uma clínica recebe pacientes durante o dia. Cada paciente possui:

- nome;
- idade;
- prioridade.

Exemplo:

```text
Ana — 32 anos — Normal
Bruno — 70 anos — Prioridade
Carlos — 45 anos — Normal
```

O sistema deve permitir:

1. adicionar paciente;
2. listar pacientes;
3. atender o primeiro paciente;
4. verificar se a fila está vazia;
5. informar a quantidade de pacientes.

---
\
RESOLUÇÃO (Feito em conjunto com João Pedro Guimarães - 42108772):

````python
# ======================================================================
# PROBLEMA REAL — Sistema de atendimento da clínica (seções 18 a 23)
# ======================================================================
class Paciente:
    """Guarda os dados de um paciente: nome, idade e prioridade."""

    def __init__(self, nome, idade, prioridade):
        self.nome = nome
        self.idade = idade
        self.prioridade = prioridade  # "Normal" ou "Prioridade"

    def __str__(self):
        return f"{self.nome} — {self.idade} anos — {self.prioridade}"


class Node:
    """Nó da fila: guarda um paciente e a referência para o próximo nó."""

    def __init__(self, paciente):
        self.paciente = paciente
        self.proximo = None


class FilaAtendimento:
    """
    Fila encadeada com início e fim.
    Pacientes normais entram no fim (FIFO). Pacientes com prioridade
    "furam a fila", mas só até o último prioritário já presente —
    ou seja, mantêm a ordem de chegada dentro do próprio grupo
    (seção 23, desafio de atendimento prioritário).
    """

    def __init__(self):
        self.inicio = None
        self.fim = None
        self._tamanho = 0

    def esta_vazia(self):
        return self.inicio is None

    def tamanho(self):
        return self._tamanho

    def adicionar(self, paciente):
        novo = Node(paciente)

        if self.inicio is None:
            # Fila vazia: o novo nó é ao mesmo tempo início e fim.
            self.inicio = novo
            self.fim = novo

        elif paciente.prioridade == "Prioridade":
            if self.inicio.paciente.prioridade != "Prioridade":
                # Não há nenhum prioritário na frente -> vai para o início.
                novo.proximo = self.inicio
                self.inicio = novo
            else:
                # Anda até o último prioritário em sequência, para inserir
                # logo depois dele (preservando ordem de chegada entre
                # prioritários) e antes do primeiro paciente normal.
                atual = self.inicio
                while (atual.proximo is not None
                       and atual.proximo.paciente.prioridade == "Prioridade"):
                    atual = atual.proximo
                novo.proximo = atual.proximo
                atual.proximo = novo
                if novo.proximo is None:
                    self.fim = novo

        else:
            # Paciente normal: sempre vai para o fim da fila.
            self.fim.proximo = novo
            self.fim = novo

        self._tamanho += 1

    def atender(self):
        """Remove e retorna o paciente do início da fila (FIFO)."""
        if self.esta_vazia():
            return None

        paciente = self.inicio.paciente
        self.inicio = self.inicio.proximo
        if self.inicio is None:
            # A fila ficou vazia -> 'fim' também precisa voltar a None.
            # (esquecer esta linha foi exatamente o bug da seção de revisão)
            self.fim = None
        self._tamanho -= 1
        return paciente

    def listar(self):
        """Imprime os pacientes na ordem em que serão atendidos."""
        if self.esta_vazia():
            print("(fila vazia)")
            return
        atual = self.inicio
        while atual is not None:
            print(atual.paciente)
            atual = atual.proximo


def demo_sistema_atendimento():
    print("--- Seção 22: uso básico ---")
    fila = FilaAtendimento()
    fila.adicionar(Paciente("Ana", 32, "Normal"))
    fila.adicionar(Paciente("Bruno", 70, "Prioridade"))
    fila.adicionar(Paciente("Carlos", 45, "Normal"))

    fila.listar()

    paciente = fila.atender()
    print("Atendendo:", paciente.nome)

    print("\n--- Seção 23: desafio de atendimento prioritário ---")
    fila2 = FilaAtendimento()
    fila2.adicionar(Paciente("Ana", 30, "Normal"))
    fila2.adicionar(Paciente("Bruno", 40, "Normal"))
    fila2.adicionar(Paciente("Carlos", 50, "Prioridade"))

    print("Fila resultante (ordem de atendimento):")
    fila2.listar()

    print("\nOrdem real ao chamar atender() sucessivamente:")
    ordem = []
    while not fila2.esta_vazia():
        ordem.append(fila2.atender().nome)
    print(" -> ".join(ordem))  # esperado: Carlos -> Ana -> Bruno


# ======================================================================
# EXERCÍCIO DE REVISÃO (seção 25)
# Erro real encontrado + versão final testada com 5 pacientes
# ======================================================================
class FilaAtendimentoBugada:
    """
    Versão com o erro proposital (mantida só para fins de demonstração).

    Roteiro de depuração da seção 14 aplicado a este bug:
      1) Qual variável está incorreta?   -> self.fim, em atender()
      2) Qual é o valor dela?             -> continua apontando para o nó
                                              já removido, em vez de None
      3) Qual objeto ela referencia?      -> um nó fora da fila
      4) Essa referência deveria mudar?   -> sim, quando a fila esvazia
      5) Condição que resolve o problema? -> resetar self.fim quando
                                              self.inicio virar None
    """

    def __init__(self):
        self.inicio = None
        self.fim = None
        self._tamanho = 0

    def esta_vazia(self):
        return self.inicio is None

    def tamanho(self):
        return self._tamanho

    def adicionar(self, paciente):
        novo = Node(paciente)
        if self.fim is None:          # BUG: deveria checar self.inicio
            self.inicio = novo
            self.fim = novo
        else:
            self.fim.proximo = novo
            self.fim = novo
        self._tamanho += 1

    def atender(self):
        if self.esta_vazia():
            return None
        paciente = self.inicio.paciente
        self.inicio = self.inicio.proximo
        # BUG: falta resetar self.fim quando a fila esvazia aqui
        self._tamanho -= 1
        return paciente

    def listar(self):
        if self.esta_vazia():
            print("(fila vazia)")
            return
        atual = self.inicio
        while atual is not None:
            print(atual.paciente)
            atual = atual.proximo


def demonstrar_bug():
    print("--- Reproduzindo o erro real ---")
    fila = FilaAtendimentoBugada()
    fila.adicionar(Paciente("Ana", 32, "Normal"))
    fila.atender()  # esvazia a fila, mas self.fim fica "preso" no nó antigo
    fila.adicionar(Paciente("Bruno", 45, "Normal"))

    print("tamanho():", fila.tamanho())        # mostra 1
    print("esta_vazia():", fila.esta_vazia())  # mostra True -> inconsistente!
    print("listar():")
    fila.listar()  # imprime "(fila vazia)" mesmo com tamanho() == 1
    print(">> tamanho diz 1, mas a fila 'parece' vazia: bug confirmado.\n")


def testar_versao_corrigida():
    print("--- Versão corrigida, testada com 5 pacientes ---")
    fila = FilaAtendimento()

    print("Fila recém-criada. Vazia?", fila.esta_vazia(), "| tamanho:", fila.tamanho())

    fila.adicionar(Paciente("Ana", 32, "Normal"))
    print("Após 1 paciente. Vazia?", fila.esta_vazia(), "| tamanho:", fila.tamanho())

    fila.adicionar(Paciente("Bruno", 70, "Prioridade"))
    fila.adicionar(Paciente("Carlos", 45, "Normal"))
    fila.adicionar(Paciente("Diana", 8, "Prioridade"))
    fila.adicionar(Paciente("Eduardo", 50, "Normal"))

    print("\nFila com 5 pacientes (ordem de atendimento):")
    fila.listar()
    print("tamanho:", fila.tamanho())

    print("\nAtendendo todos, um a um:")
    while not fila.esta_vazia():
        p = fila.atender()
        print("Atendendo:", p.nome, "| restam:", fila.tamanho())

    print("\nFila esvaziada. Vazia?", fila.esta_vazia(), "| tamanho:", fila.tamanho())

    print("\nRepetindo o cenário do bug (adicionar após esvaziar):")
    fila.adicionar(Paciente("Fernanda", 28, "Normal"))
    print("tamanho():", fila.tamanho())
    print("esta_vazia():", fila.esta_vazia())
    fila.listar()  # agora mostra Fernanda corretamente



if __name__ == "__main__":

    print("\n=== Sistema de atendimento da clínica (seções 18-23) ===")
    demo_sistema_atendimento()

    print("\n=== Exercício de revisão (seção 25) ===")
    demonstrar_bug()
    testar_versao_corrigida()
````

## 27. Reflexão final

### O que é uma referência em Python?

```text
É um ponteiro ou endereço interno que conecta um nome (variável) a um objeto armazenado na memória
```

### Qual a diferença entre `b = a` e `b = a.copy()`?

```text
b = a aponta ambas as variáveis para o mesmo objeto na memória, enquanto b = a.copy() cria um novo objeto, fazendo os dois terem id's diferentes.
```

### Por que uma estrutura encadeada precisa de referências?

```text
Porque seus elemtnos não ficam em blocos contínuos na memória e sim espalhados, então é necessário ter uma "corrente" pro dado saber qual é o próximo.
```
