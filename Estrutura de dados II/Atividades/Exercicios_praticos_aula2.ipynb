## Exercício 1 — Soma de um vetor

Crie um programa que leia **10 números inteiros** e armazene-os em um array.

Ao final, apresente:

* todos os números;
* a soma dos elementos;
* a média dos valores.

````python
%%writefile soma_vetor.c

#include <stdio.h>

int main()
{
    // Declarando tamanho para fácil variação de testes
    int tamanho = 10;
    int numeros[tamanho], i;
    int resultado = 0;

    // laço for para o vetor receber os números do usuário
    for(i = 0; i < tamanho; i++) {
      printf("Digite um número para a posição %d: ", i);
      scanf("%d", &numeros[i]);
    }

    // realizando a soma
    for(i = 0; i < tamanho; i++) {
      resultado += numeros[i];
    }

    // exibindo resultado
    printf("resultado da soma dos elementos do vetor: %d", resultado);

    return 0;
}
````

## Exercício 2 — Maior e menor

Leia **10 números inteiros** utilizando um array.

Determine:

* o maior valor;
* o menor valor;
* as posições em que eles aparecem.

````python
%%writefile maior_menor.c

#include <stdio.h>

int main()
{
    // Declarando tamanho para fácil variação de testes / declarando maior e menor
    int tamanho = 10;
    int numeros[tamanho], i;
    int maior = numeros[0];
    int menor = numeros[0];

    // laço for para o vetor receber os números do usuário
    for(i = 0; i < tamanho; i++) {
      printf("Digite um número para a posição %d: ", i);
      scanf("%d", &numeros[i]);
    }

    // verificando maior elemento
    for (i = 1; i < tamanho; i++) {
        if (numeros[i] > maior) {
            maior = numeros[i];
        }
    }

    // verificando menor elemento
    for (i = 1; i < tamanho; i++) {
        if (numeros[i] < menor) {
            menor = numeros[i];
        }
    }


    // exibindo valores
    printf("Maior elemento: %d\n", maior);
    printf("Menor elemento: %d\n", menor);


    // exibindo posição dos elementos
    for (i = 0; i < tamanho; i++) {
        printf("Elemento na posição %d: %d\n", i, numeros[i]);
    }

    return 0;
}

````

## Exercício 3 — Números pares

Leia **20 números inteiros** e armazene-os em um array.

Depois:

* mostre somente os números pares;
* conte quantos números pares existem;
* calcule a soma dos números pares.

````python
%%writefile numeros_pares.c

#include <stdio.h>

int main()
{
    // Declarando tamanho para fácil variação de testes / inicializando varíavel de soma
    int tamanho = 20;
    int numeros[tamanho], i;
    int resultado = 0;
    int contabilizacao = 0;

    // laço for para o vetor receber os números do usuário
    for(i = 0; i < tamanho; i++) {
      printf("Digite um número para a posição %d: ", i);
      scanf("%d", &numeros[i]);
    }

    // contabilizando números pares
    for(i = 0; i < tamanho; i++) {
        if (numeros[i] % 2 == 0) {
          contabilizacao += 1;
        }
    }

    // somando números pares
    for(i = 0; i < tamanho; i++) {
        if (numeros[i] % 2 == 0) {
          resultado += numeros[i];
        }
    }

    // exibindo informações
    printf("\nresultado da soma dos elementos pares: %d\n", resultado);
    printf("quantidade de números pares: %d\n", contabilizacao);

    for(i = 0; i < tamanho; i++) {
        if (numeros[i] % 2 == 0) {
          printf(" %d ", numeros[i]);
        }
    }

  return 0;
}
````

## Exercício 4 — Inversão de vetor

Leia **10 números** e armazene-os em um array.

Mostre o vetor original e o vetor invertido.

````python
%%writefile inversao_vetor.c

#include <stdio.h>

int main()
{
    // Declarando tamanho para fácil variação de testes
    int tamanho = 10;
    int numeros[tamanho], i;
    int numeros_invertidos[tamanho];

    // laço for para o vetor receber os números do usuário
    for(i = 0; i < tamanho; i++) {
      printf("Digite um número para a posição %d: ", i);
      scanf("%d", &numeros[i]);
    }

    // exibindo vetor original
    for(i = 0; i < tamanho; i++) {
        printf(" %d ", numeros[i]);
    }

    printf("\n");

     // Copia os elementos de trás para frente
    for (i = 0; i < tamanho; i++) {
        numeros_invertidos[i] = numeros[tamanho - 1 - i];
    }

    printf("\n");

    // exibindo valor invertido
    for(i = 0; i < tamanho; i++) {
        printf(" %d ", numeros_invertidos[i]);
    }


}
````

## Exercício 5 — Matriz 3 × 3

Crie uma matriz `3 x 3` de números inteiros.

O programa deverá:

* ler os valores;
* mostrar a matriz;
* calcular a soma de todos os elementos;
* mostrar o maior valor.


````python
%%writefile matriz_3x3.c

#include <stdio.h>

int main()
{
    int tamanho_i = 3;
    int tamanho_j = 3;
    int matriz[tamanho_i][tamanho_j], i, j;
    int soma = 0;
    int maior = matriz[0][0];


    // laço for para a matriz receber os números do usuário / linha e coluna, for aninhados
    for(i = 0; i < tamanho_i; i++) {
      for(j = 0; j < tamanho_j; j++) {
        printf("Digite um número para a posição [%d][%d]: ", i, j);
        scanf("%d", &matriz[i][j]);
      }
    }

    // exibição da matriz
    printf("\nMatriz: \n");

    for(i = 0; i < tamanho_i; i++) {
      for(j = 0; j < tamanho_j; j++) {
        printf(" %d ", matriz[i][j]);
      }
      printf("\n");
    }


    // exibindo soma dos elementos da matriz
    for(i = 0; i < tamanho_i; i++) {
      for(j = 0; j < tamanho_j; j++) {
        soma += matriz[i][j];
      }
      printf("\n");
    }

    printf("Soma da matriz: %d\n", soma);


    // exibindo o maior valor / se o valor for maior que o do primeiro elemento, ele vira o maior, e assim percorrendo a matriz
    for(i = 0; i < tamanho_i; i++) {
      for(j = 0; j < tamanho_j; j++) {
        if (matriz[i][j] > maior) {
            maior = matriz[i][j];
        }
      }
      printf("\n");
    }

    printf("Maior elemento do vetor %d", maior);

}
````

## Exercício 6 — Diagonal principal

Leia uma matriz `4 x 4`.

Mostre os elementos da **diagonal principal** e calcule sua soma.

````python
%%writefile diagonal_principal.c

#include <stdio.h>

int main()
{
    int tamanho_i = 4;
    int tamanho_j = 4;
    int matriz[tamanho_i][tamanho_j], i, j;
    int soma = 0;

    // laço for para a matriz receber os números do usuário / linha e coluna, for aninhados
    for(i = 0; i < tamanho_i; i++) {
      for(j = 0; j < tamanho_j; j++) {
        printf("Digite um número para a posição [%d][%d]: ", i, j);
        scanf("%d", &matriz[i][j]);
      }
    }

    // exibindo matriz inteira primeiro

    printf("\nMatriz: \n");

    for(i = 0; i < tamanho_i; i++) {
      for(j = 0; j < tamanho_j; j++) {
        printf(" %d ", matriz[i][j]);
      }
      printf("\n");
    }

    printf("\n");

    printf("Elementos da diagonal: \n");
    // exibindo diagnonal da matriz / i e j são iguais nesse caso e é utilizado apenas um for
    for(i = 0; i < tamanho_i; i++) {
      printf(" %d ", matriz[i][i]);
    }


    // somando diagonal da matriz
    for(i = 0; i < tamanho_i; i++) {
      soma += matriz[i][i];
    }

    printf("\n resultado da soma dos elementos da diagonal: %d", soma);

}
````

## Exercício 7 — Matriz de notas

Uma turma possui **4 alunos e 3 avaliações**.

Crie uma matriz para armazenar as notas:

```text
          P1   P2   P3
Aluno 1
Aluno 2
Aluno 3
Aluno 4
```

Calcule e apresente a média de cada aluno.


````python
%%writefile medias_alunos.c

#include <stdio.h>

int main()
{
    // dessa vez a variável soma é um float por serem notas de alunos
    int tamanho_i = 4;
    int tamanho_j = 3;
    int matriz[tamanho_i][tamanho_j], i, j;


    // laço for para a matriz receber os números do usuário / linha e coluna, for aninhados
    for(i = 0; i < tamanho_i; i++) {
      printf("Digite as 3 notas do Aluno %d: ", i + 1);

      for(j = 0; j < tamanho_j; j++) {
        scanf("%d", &matriz[i][j]);
      }
    }


    // exibindo matriz para facil visualização das notas
    printf("\nMatriz: \n");

    for(i = 0; i < tamanho_i; i++) {
      for(j = 0; j < tamanho_j; j++) {
        printf(" %d ", matriz[i][j]);
      }
      printf("\n");
    }

    printf("\n");


    // realizando calculo de média das notas / primeiro a soma, depois a divisão
    for(i = 0; i < tamanho_i; i++) {
      float soma = 0;

      for(j = 0; j < tamanho_j; j++) {
        soma += matriz[i][j];
      }


      // aqui é usado o tamanho_j pois são 3 "avaliações"
      float media = soma / tamanho_j;
      printf("Aluno %d - Média: %.2f\n", i + 1, media);
    }


}
````

## Exercício 8 — Struct Produto

Crie uma `struct Produto` contendo:

```text
nome
codigo
preco
quantidade
```

Cadastre **cinco produtos**.

Depois:

* mostre todos os produtos;
* calcule o valor total de cada produto;
* informe o produto com maior valor em estoque.

### Fórmula

```text
valor em estoque = preço × quantidade
```

````python
%%writefile produtos.c

#include <stdio.h>

int main()
{
  // declarando quantos produtos serão inseridos no sistema
  int i, total_produtos = 5;
  int maior = 0;
  float valor_estoque = 0;

  // criando a struct
  struct Produto {
    char nome[50];
    int codigo;
    float preco;
    int quantidade;
  };

  // formando uma array de structs
  struct Produto produtos[5];

  printf("- CADASTRO DE PRODUTOS -");

  for(i = 0; i < total_produtos; i++) {
    printf("\nDigite o nome do produto: ");
    scanf("%s", &produtos[i].nome);

    printf("\nDigite o código do produto: ");
    scanf("%d", &produtos[i].codigo);

    printf("\nDigite o preco do produto: ");
    scanf("%f", &produtos[i].preco);

    printf("\nDigite a quantidade do produto: ");
    scanf("%d", &produtos[i].quantidade);
  }

  // exibindo produtos
  printf("\nPRODUTOS:\n");
  for(i = 0; i < total_produtos ; i++) {
    printf("Nome: %s | Código: %d | Preço: %.2f | Quantidade: %d \n", produtos[i].nome, produtos[i].codigo, produtos[i].preco, produtos[i].quantidade);
  }

  // calculando valor de estoque
  printf("\nVALOR DE ESTOQUE:\n");
  for(i = 0; i < total_produtos ; i++) {
    valor_estoque = produtos[i].preco * produtos[i].quantidade;

    printf("\nNome: %s | valor de estoque: %.2f", produtos[i].nome, valor_estoque);
  }


  printf("\nMAIOR VALOR DO ESTOQUE");
  // informando produto com maior valor
  // pega o valor de estoque do primeiro produto para usar como base
  float maior_valor_estoque = produtos[0].preco * produtos[0].quantidade;
  maior = 0;

  for (i = 1; i < total_produtos; i++) {
    float valor_atual = produtos[i].preco * produtos[i].quantidade;

    if (valor_atual > maior_valor_estoque) {
        maior_valor_estoque = valor_atual;
        maior = i;
    }
  }

  printf("\n%s | %.2f", produtos[maior].nome, maior_valor_estoque);


}
````

## Exercício 9 — Struct Aluno

Crie uma `struct Aluno` contendo:

```text
nome
idade
nota1
nota2
nota3
```

Cadastre **cinco alunos**.

Calcule a média de cada aluno e classifique:

```text
Média >= 7,0 → Aprovado
Média < 7,0  → Reprovado
```

Ao final, informe:

* quantidade de aprovados;
* quantidade de reprovados;
* aluno com maior média.


````python
%%writefile struct_aluno.c

#include <stdio.h>

int main()
{
  int i, total_alunos = 5;
  int soma = 0;

  // criando a struct
  struct Aluno {
    char nome[50];
    int idade;
    float nota1;
    float nota2;
    float nota3;
  };

  // formando uma array de structs
  struct Aluno alunos[5];


  for(i = 0; i < total_alunos; i++) {
    printf("\nDigite o nome do aluno: ");
    scanf("%s", &alunos[i].nome);

    printf("\nDigite a idade do aluno: ");
    scanf("%d", &alunos[i].idade);

    printf("\nDigite a 1ª nota do aluno: ");
    scanf("%f", &alunos[i].nota1);

    printf("\nDigite a 2ª nota do aluno: ");
    scanf("%f", &alunos[i].nota2);

    printf("\nDigite 3ª nota do aluno: ");
    scanf("%f", &alunos[i].nota3);
  }


    for(i = 0; i < total_alunos; i++) {
      float media = (alunos[i].nota1 + alunos[i].nota2 + alunos[i].nota3) / 3.0;

      printf("\nAluno: %s\n", alunos[i].nome);
      printf("Idade: %d\n", alunos[i].idade);
      printf("Notas: %.1f | %.1f | %.1f\n", alunos[i].nota1, alunos[i].nota2, alunos[i].nota3);
      printf("Média: %.2f\n", media);

      // verificando situação de aprovação
      if (media >= 6.0) {
          printf("Status: Aprovado\n");
        } else {
          printf("Status: Reprovado\n");
      }
    }




}
````

# Exercício 10 — Sistema Integrado

Crie um pequeno sistema para cadastro de **10 funcionários**.

Utilize uma `struct` contendo:

```text
nome
idade
cargo
salario
```

O programa deverá permitir:

1. Cadastrar os funcionários.
2. Listar todos os funcionários.
3. Mostrar o funcionário com maior salário.
4. Calcular a média salarial.
5. Mostrar os funcionários com salário acima da média.

````python
%%writefile sistema_integrado.c

#include <stdio.h>

int main()
{
  int i, total_funcionarios = 10;
  int maior = 0;
  float soma_salarios = 0.0;
  float media_salarial = 0.0;

  // criando a struct
  struct Funcionario {
    char nome[50];
    int idade;
    char cargo[50];
    float salario;
  };

  // formando uma array de structs
  struct Funcionario funcionarios[10];

  for(i = 0; i < total_funcionarios; i++) {

    printf("\nDigite o nome do funcionario: ");
    scanf("%s", &funcionarios[i].nome);

    printf("\nDigite a idade: ");
    scanf("%d", &funcionarios[i].idade);

    printf("\nDigite o cargo do funcionario: ");
    scanf("%s", &funcionarios[i].cargo);

    printf("\nDigite o salário do funcionario: ");
    scanf("%f", &funcionarios[i].salario);

    // aproveitando para somar os salários logo depois das entradas
    soma_salarios += funcionarios[i].salario;
  }

  // Calculando a média salarial
  media_salarial = soma_salarios / total_funcionarios;

  // Listando funcionarios
  printf("\nFUNCIONÁRIOS:\n");
  for(i = 0; i < total_funcionarios ; i++) {
    printf("Nome: %s | Idade: %d | Cargo: %s | Salario: %.2f \n", funcionarios[i].nome, funcionarios[i].idade, funcionarios[i].cargo, funcionarios[i].salario);
  }

  // Exibindo o maior salário
  for(i = 0; i < total_funcionarios ; i++) {
    if (funcionarios[i].salario > funcionarios[maior].salario) {
      maior = i;
    }
  }
  printf("\nMAIOR SALÁRIO:\n");
  printf("Nome: %s | Salário: %.2f\n", funcionarios[maior].nome, funcionarios[maior].salario);


  // Exibindo a média salarial
  printf("\nMÉDIA SALARIAL:\n");
  printf("R$ %.2f\n", media_salarial);


  // Exibindo funcionários com salário acima da média
  printf("\nFUNCIONÁRIOS COM SALÁRIO ACIMA DA MÉDIA:\n");
  int encontrou = 0;
  for(i = 0; i < total_funcionarios; i++) {
    if (funcionarios[i].salario > media_salarial) {
      printf("Nome: %s | Salário: %.2f\n", funcionarios[i].nome, funcionarios[i].salario);
      encontrou = 1;
    }
  }

  if (encontrou != 1) {
    printf("Nenhum funcionário ganha acima da média.\n");
  }



}
````

## Desafio extra

Crie um menu:

```text

```


Usarei o mesmo código da atividade 10, mas agora com o menu interativo

````python
%%writefile menu_interativo.c

#include <stdio.h>

int main()
{

  // aqui foi adicionado a variável opção pra tornar o menu funcional e possível de existir
  int i, total_funcionarios = 2;
  int maior = 0;
  int opcao;
  float soma_salarios = 0.0;
  float media_salarial = 0.0;

  // criando a struct
  struct Funcionario {
    char nome[50];
    int idade;
    char cargo[50];
    float salario;
  };

  // formando uma array de structs
  struct Funcionario funcionarios[2];


  printf("-- CADASTRO DE FUNCIONARIO --");
  for(i = 0; i < total_funcionarios; i++) {
    printf("\nFuncionario %d:\n", i + 1);

    printf("\nDigite o nome do funcionario: ");
    scanf("%s", &funcionarios[i].nome);

    printf("\nDigite a idade: ");
    scanf("%d", &funcionarios[i].idade);

    printf("\nDigite o cargo do funcionario: ");
    scanf("%s", &funcionarios[i].cargo);

    printf("\nDigite o salário do funcionario: ");
    scanf("%f", &funcionarios[i].salario);

    // aproveitando para somar os salários logo depois das entradas
    soma_salarios += funcionarios[i].salario;
  }

  // Calculando a média salarial
  media_salarial = soma_salarios / total_funcionarios;

  // menu:
  do {
    printf("\n-- MENU INTERATIVO --\n");
    printf("\n1. Listar todos os funcionários");
    printf("\n2. Exibir o funcionário com maior salário");
    printf("\n3. Exibir a média salarial");
    printf("\n4. Mostrar funcionários acima da média");
    printf("\n5. Sair");
    printf("\n==============================");
    printf("\nEscolha uma opção: ");
    scanf("%d", &opcao);

    switch(opcao) {
      case 1:
        printf("\nLISTA DE FUNCIONÁRIOS\n");
        for(i = 0; i < total_funcionarios; i++) {
          printf("Nome: %s | Idade: %d | Cargo: %s | Salario: %.2f\n",
                 funcionarios[i].nome, funcionarios[i].idade, funcionarios[i].cargo, funcionarios[i].salario);
        }
        break;


      case 2:
        int indice_maior = 0;
        for(i = 0; i < total_funcionarios; i++) {
          if (funcionarios[i].salario > funcionarios[maior].salario) {
            maior = i;
          }
        }
        printf("\n--- MAIOR SALÁRIO ---\n");
        printf("Nome: %s | Salário: %.2f\n", funcionarios[maior].nome, funcionarios[maior].salario);
        break;


      case 3:
        printf("\n--- MÉDIA SALARIAL ---\n");
        printf("R$ %.2f\n", media_salarial);
        break;


      case 4:
        printf("\n--- ACIMA DA MÉDIA ---\n");
        int encontrou = 0;
        for(i = 0; i < total_funcionarios; i++) {
          if (funcionarios[i].salario > media_salarial) {
            printf("Nome: %s | Salário: %.2f\n", funcionarios[i].nome, funcionarios[i].salario);
            encontrou = 1;
          }
        }
        if (encontrou != 1) {
          printf("Nenhum funcionário ganha acima da média.\n");
        }
        break;


      case 5:
        printf("\nEncerrando o sistema...\n");
        break;


      default:
        printf("\nOpção inválida! Tente novamente.\n");
    }
  } while (opcao != 5);
}
````
